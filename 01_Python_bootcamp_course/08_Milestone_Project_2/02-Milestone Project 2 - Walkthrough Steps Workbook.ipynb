## Game Play
To play a hand of Blackjack the following steps must be followed:
1. Create a deck of 52 cards
2. Shuffle the deck
3. Ask the Player for their bet
4. Make sure that the Player's bet does not exceed their available chips
5. Deal two cards to the Dealer and two cards to the Player
6. Show only one of the Dealer's cards, the other remains hidden
7. Show both of the Player's cards
8. Ask the Player if they wish to Hit, and take another card
9. If the Player's hand doesn't Bust (go over 21), ask if they'd like to Hit again.
10. If a Player Stands, play the Dealer's hand. The dealer will always Hit until the Dealer's value meets or exceeds 17
11. Determine the winner and adjust the Player's chips accordingly
12. Ask the Player if they'd like to play again

## Playing Cards
A standard deck of playing cards has four suits (Hearts, Diamonds, Spades and Clubs) and thirteen ranks (2 through 10, then the face cards Jack, Queen, King and Ace) for a total of 52 cards per deck. Jacks, Queens and Kings all have a rank of 10. Aces have a rank of either 11 or 1 as needed to reach 21 without busting. **Step 2: Create a Card Class**<br>
A Card object really only needs two attributes: suit and rank. You might add an attribute for "value" - we chose to handle value later when developing our Hand class.<br>In addition to the Card's \_\_init\_\_ method, consider adding a \_\_str\_\_ method that, when asked to print a Card, returns a string in the form "Two of Hearts"

```python
class Card:
    def __init__(self, rank, suit):
        self.rank = rank
        self.suit = suit
    
    def __str__(self):
        return f"{self.rank} of {self.suit}"

    def __repr__(self):
        return self.__str__()
```

**Step 3: Create a Deck Class**<br>
Here we might store 52 card objects in a list that can later be shuffled. First, though, we need to *instantiate* all 52 unique card objects and add them to our list. So long as the Card class definition appears in our code, we can build Card objects inside our Deck \_\_init\_\_ method. Consider iterating over sequences of suits and ranks to build out each card. This might appear inside a Deck class \_\_init\_\_ method:

In addition to an \_\_init\_\_ method we'll want to add methods to shuffle our deck, and to deal out cards during gameplay.<br><br>
OPTIONAL: We may never need to print the contents of the deck during gameplay, but having the ability to see the cards inside it may help troubleshoot any problems that occur during development. With this in mind, consider adding a \_\_str\_\_ method to the class definition.

```python
class Deck:
    
    def __init__(self):
        self.deck = []  # start with an empty list
        for suit in suits:
            for rank in ranks:
                self.deck.append(Card(rank, suit))
    
    def __str__(self):
        deck_content = "The deck contains\n"
        
        for i, card in enumerate(self.deck):
            deck_content += f"  {card.rank} of {card.suit}\n"

        return deck_content
 
    def shuffle(self):
        random.shuffle(self.deck)
        
    def deal(self):
        if len(self.deck) == 0:
            return "The deck is finish. End of game"

        single_player_card = self.deck.pop(0)
        return single_player_card
```

**Step 6: Write a function for taking bets**<br>
Since we're asking the user for an integer value, this would be a good place to use <code>try</code>/<code>except</code>. Remember to check that a Player's bet can be covered by their available chips.

```python
def take_bet(player, dealer, dealer_reserverd):
    
    while True:
        if player.chips.total <= 0:
            print(f"{player.name} has no chips left.")
            player.place_a_bet = False
            player.chips.bet = 0
            return 0

        place_bet = input(f"Would {player.name} like to place a bet: ").lower()

        if place_bet != "yes":
            player.chips.bet = 0
            player.place_a_bet = False
            return 0
        
        player.place_a_bet = True

        available = dealer.chips.total - dealer_reserverd
        if available <= 0:
            print("Dealer can't accept any more bets")
            player.place_a_bet = False
            player.chips.bet = 0
            return 0
        
        try:
            player.chips.bet = int(input("Give your bet: "))

            if player.chips.bet <= 0:
                print("Bets must be greater than 0")
                continue

            if player.chips.bet > available:
                print("The dealer can't match your bet")
                print("The player can match the max bet of the dealer")

                player_match_bet = input(f"Would {player.name} match dealers max bet chips").lower()
                if player_match_bet == "yes":
                    player.chips.bet = available
                    
        except ValueError:
            print("Sorry a bet must be an integer")
        else:
            
            if player.chips.bet >= player.chips.total and player.chips.total > 0:
                print("A player is going for an all in")
                player.chips.bet = player.chips.total

            if player.chips.bet > available:
                print(f"Dealer can only cover {available} chips")
                
                match_bet = input(f"Would player {player.name} like to match the available max bet of {available} chips").lower()
                if match_bet == "yes":
                    player.chips.bet = available

                else:
                    continue
            
            print("A bet has been placed")
            player.chips.total -= player.chips.bet

            return player.chips.bet
```

**Step 7: Write a function for taking hits**<br>

**Step 8: Write a function prompting the Player to Hit or Stand**<br>
This function should accept the deck and the player's hand as arguments, and assign playing as a global variable.<br>
If the Player Hits, employ the hit() function above. If the Player Stands, set the playing variable to False - this will control the behavior of a <code>while</code> loop later on in our code.

```python
def hit_or_stand(deck,hand):
    while True:
        if hand.name != "Dealer":
            if hand.value > 21:
                return "Player busted"
            
            print(f"{hand.name} has cards: {hand}")
            player_choice = input(f"Player {hand.name} turn to hit or stand").upper()

            if player_choice == "HIT":
                hit(deck, hand)

            elif player_choice == "STAND":
                print(f"Player {hand.name} stands. Next player's turn")
                break

            else:
                print("Please type hit or stand")
                continue
        else:
            if hand.value < 17:
                hit(deck, hand)
            else:
                print("Dealer stands.")
                break
```

**Step 10: Write functions to handle end of game scenarios**<br>
Remember to pass player's hand, dealer's hand and chips as needed.

```python
class Table:
    def __init__(self, name, table_chips = 10000, player_limit = 6):
        self.name = name
        self.player_limit = player_limit
        self.dealer = Players("Dealer", table_chips)
        self.players = []
        self.players_at_the_table = False

        self.players.append(self.dealer)

    def give_chips_to_players(self, amount):
        self.dealer.chips.total -= amount

    def add_players(self, players):
        if isinstance(players, (list, tuple)):
            for player in players:
                accept_player = input(f"Should the player {player.name} be accepted").lower()
                
                if self.player_limit == 0:
                    print("Cannot exceed player limit")
                    break

                if accept_player == "yes":
                    self.players_at_the_table = True

                    self.players.append(player)
                    self.give_chips_to_players(player.chips.total)
                    
                    self.player_limit -= 1
                else:
                    f"Player {player.name} was declined"
        else:
            self.players_at_the_table = True
            self.players.append(players)
            self.give_chips_to_players(players.chips.total)

    def player_wins(self, player):
        player.chips.win_bet()
        self.dealer.chips.total -= player.chips.bet

    def player_wins_with_blackjack(self, player):
        payout = player.chips.bet * 2.5
         
        player.chips.total += payout
        self.dealer.chips.total -= player.chips.bet * 1.5

    def player_loses(self, player):
        self.dealer.chips.total += player.chips.bet

    def push(self, player):
        player.chips.tie_bet()

    def determine_winner(self):
        for player in self.players[1:]:

            if not player.place_a_bet:
                continue

            if player.value > 21:
                print(f"{player.name} busted!")
                self.player_loses(player)

            elif self.dealer.value > 21:
                print(f"{player.name} wins! Dealer busted!")
                self.player_wins(player)

            elif player.value == 21 and len(player.cards) == 2:
                print(f"{player.name} got BLACKJAAAACK")
                self.player_wins_with_blackjack(player)

            elif player.value > self.dealer.value:
                print(f"{player.name} wins!")
                self.player_wins(player)

            elif player.value < self.dealer.value:
                print(f"{player.name} loses!")
                self.player_loses(player)

            else:
                print(f"{player.name} pushes!")
                self.push(player)

            print(player.chips.total)

        print(self.dealer.chips.total)

    def reset_after_turn(self):
        for player in self.players:
            player.cards = []
            player.value = 0
            player.chips.bet = 0
            player.aces = 0
            player.place_a_bet = False
                
    def play_blackjack(self, players):
        deck = Deck()
        deck.shuffle()
        self.add_players(players)
        turn = 1

        print("Welcome to our blackjack game!")
        print("Buy-ins are bound to the tables limits")
        print("Play with amounts that won't affect your, life permentantly, and with caution")
        print("Gambling is meant to be played for fun!!!\n")
            
        while any(player.chips.total > 0 for player in self.players[1:]):
            if self.dealer.chips.total <= 0:
                print("The casino is bankrupt!")
                print(f"Players have beaten the dealers")
                break
            
            if len(deck.deck) < 20:
                deck = Deck()
                deck.shuffle()
                print("Shuffling a new deck!")

            self.reset_after_turn()
            print(f"Turn: {turn}")
            print(f"The Dealer has: {self.dealer.chips.total} chips left")
            dealer_reserverd = 0

            for player in self.players[1:]:
                bet = take_bet(player, self.dealer, dealer_reserverd)
                dealer_reserverd += bet

                if bet > 0:
                    print(f"Player: {player.name} has placed a bet: {player.chips.bet}")

            for _ in range(2):
                for player in self.players:
                    if player == self.dealer or player.place_a_bet:
                        player.add_card(deck.deal())

            show_some(players, self.dealer)

            for player in self.players[1:]:
                print(player)
                
                if player.place_a_bet:
                    hit_or_stand(deck, player) 

            hit_or_stand(deck, self.dealer)

            turn += 1

            show_all(players, self.dealer)
            self.determine_winner()

        else:
            print("Game over!")
            print("The Dealer has beaten the players!")
```
