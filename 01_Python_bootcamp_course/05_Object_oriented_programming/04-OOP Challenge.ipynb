# Object Oriented Programming Challenge

For this challenge, create a bank account class that has two attributes:

* owner
* balance

and two methods:

* deposit
* withdraw

As an added requirement, withdrawals may not exceed the available balance.

Instantiate your class, make several deposits and withdrawals, and test to make sure the account can't be overdrawn.

```python
class Account:
    def __init__(self, owner, balance):
        if balance < 0:
            raise ValueError("Balance cant have negative value.")

        self.owner = owner
        self.balance = balance

    def deposit(self, deposit_amount):
        self.balance += deposit_amount
        print("Deposit accepted")
        print(f"Current balance: {self.balance} after deposit")

    def withdraw(self, withdraw_amount):
        if self.balance - withdraw_amount < 0:
            raise ValueError("Can't withdraw more than the account of the balance")

        self.balance -= withdraw_amount
        print("Withdraw accepted")
        print(f"Current balance: {self.balance} after withdraw")

    def __repr__(self):
        return f"Account owner: {self.owner}\nAccount balance: {self.balance}"
```

```python
# 1. Instantiate the class
acct1 = Account('Jose',100)
```

```python
# 2. Print the object
print(acct1)
```

```python
# 3. Show the account owner attribute
acct1.owner
```

```python
# 4. Show the account balance attribute
acct1.balance
```

```python
# 5. Make a series of deposits and withdrawals
acct1.deposit(50)
acct1.balance
```

```python
acct1.withdraw(75)
```

```python
# 6. Make a withdrawal that exceeds the available balance
acct1.withdraw(500)
```
