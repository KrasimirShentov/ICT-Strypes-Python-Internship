#### Problem 1
Fill in the Line class methods to accept coordinates as a pair of tuples and return the slope and distance of the line.

```python
import math
```

```python
class Line:
    
    def __init__(self,coor1,coor2):
        self.coor1 = coor1
        self.coor2 = coor2

    def get_coords(self):
        x1, y1 = self.coor1
        x2, y2 = self.coor2

        return x1, x2, y1, y2
    
    def distance(self):
        x1, x2, y1, y2 = self.get_coords()

        return math.sqrt(((x2 - x1) ** 2 +(y2- y1) ** 2))
    
    def slope(self):
        x1, x2, y1, y2 = self.get_coords()

        return (y2 - y1) / (x2 - x1)
```

```python
# EXAMPLE OUTPUT

coordinate1 = (3,2)
coordinate2 = (8,10)

li = Line(coordinate1,coordinate2)
```

```python
li.distance()
```

```python
li.slope()
```

#### Problem 2

Fill in the class 

```python
class Cylinder:
    
    def __init__(self,height=1,radius=1):
        self.height = height
        self.radius = radius

    def volume(self):
        return math.pi * math.pow(self.radius, 2) * self.height
    
    def surface_area(self):
        bases = 2 * (math.pi * self.radius ** 2)
        curved_surface = 2 * math.pi * self.radius * self.height
        
        return bases + curved_surface
```

```python
# EXAMPLE OUTPUT
c = Cylinder(2,3)
```

```python
c.volume()
```

```python
c.surface_area()
```
