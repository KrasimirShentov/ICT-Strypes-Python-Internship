# Sorting Algorithms

```python
import random
import time
import pandas as pd
```

```python
numbers = random.sample(range(1, 10001), 10000)
print(numbers)
num_checking = random.sample(range(1, 10001), 10000)
num_checking.sort()
print(num_checking)

algorithms_results = dict()
```

```python
#Bubble sort
bubble_numbers = random.sample(range(1, 10001), 10000)
n = len(bubble_numbers)

bubble_sort_time_start = time.time()
for i in range(n):
    for j in range(n - i - 1):
        if bubble_numbers[j] > bubble_numbers[j + 1]:
            bubble_numbers[j], bubble_numbers[j + 1] = bubble_numbers[j + 1], bubble_numbers[j]

bubble_sort_time_end = time.time()
bubble_sort_time = bubble_sort_time_end - bubble_sort_time_start
algorithms_results[0] = bubble_sort_time

# print(bubble_numbers)
print(bubble_numbers == num_checking)
```

```python
#Merge sort
merge_numbers = random.sample(range(1, 10001), 10000)
merged = []

def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    
    left_half = merge_sort(arr[:mid])
    right_half = merge_sort(arr[mid:])

    return merge_arrays(left_half, right_half)

def merge_arrays(left_half, right_half):
    merged = []
    i = j = 0

    while i < len(left_half) and j < len(right_half):
        if left_half[i] <= right_half[j]:
            merged.append(left_half[i])
            i += 1
        else:
            merged.append(right_half[j])
            j += 1

    merged.extend(left_half[i:])
    merged.extend(right_half[j:])

    return merged

merge_sort_time_start = time.time()
sorted_array = merge_sort(merge_numbers)
merge_sort_time_end = time.time()

merge_sort_time = merge_sort_time_end - merge_sort_time_start
algorithms_results[1] = merge_sort_time

# print(sorted_array)
print(sorted_array == num_checking)
```

```python
#Quicksort
quicksort_numbers = random.sample(range(1, 10001), 10000)

def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        if arr[j] < pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


def quicksort(arr, low = 0, high = None):
    if high is None:
        high = len(arr) - 1

    if low < high:
        partio = partition(arr, low, high)
        quicksort(arr, low, partio - 1)
        quicksort(arr, partio + 1, high)

quicksort_time_start = time.time()
quicksort(quicksort_numbers)
quicksort_time_end = time.time()

quicksort_time = quicksort_time_end - quicksort_time_start
algorithms_results[2] = quicksort_time

# print(quicksort_numbers)
print(quicksort_numbers == num_checking)
```

```python
#Comb sort

comb_sort_numbers = random.sample(range(1, 10001), 10000)

def comb_sort(arr):
    gap = len(arr)
    is_swapped = True

    while gap != 1 or is_swapped:
        gap = int(gap / 1.3)

        if gap < 1:
            gap = 1
        is_swapped = False

        for i in range(len(arr) - gap):
            # print(arr[i], arr[i + gap])

            if arr[i] > arr[i + gap]:
                arr[i], arr[i + gap] = arr[i + gap], arr[i]

                # print(arr[i], arr[i + gap])
                is_swapped = True

comb_sort_time_start = time.time()
comb_sort(comb_sort_numbers)
comb_sort_time_end = time.time()

comb_sort_time = quicksort_time_end - comb_sort_time_start
algorithms_results[3] = comb_sort_time

# print(comb_sort_numbers)
print(comb_sort_numbers == num_checking)
```

```python
#insertion sort
insertion_sort_numbers = random.sample(range(1, 10001), 10000)

def insertions_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

insertion_sort_time_start = time.time()
insertions_sort(insertion_sort_numbers)
insertion_sort_time_end = time.time()

insertion_sort_time = insertion_sort_time_end - insertion_sort_time_start
algorithms_results[4] = insertion_sort_time

# print(insertion_sort_numbers)
print(insertion_sort_numbers == num_checking)
```
