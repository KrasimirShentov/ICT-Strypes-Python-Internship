### Regex pattern and test

```python
pattern = r"[\d]{3}-[\d]{3}-[\d]{4}"
text = "some text with a phone number 000-000-0000"

reg = re.findall(pattern, text)
print(reg)
```

```python
def regex_search(file, pattern):
    f = open(file, "r")
    text = f.read()

    if re.search(pattern, text):
        return re.search(pattern, text)
    else:
        return "no phone numbers"
```

```python
results = []

for folder, subfolder, files in os.walk(os.getcwd() + r"\extracted_content"):

    for f in files:
        file_path = os.path.join(folder, f)
        results.append(regex_search(file_path, pattern))
print(results)
```

```python
for result in results:
    if result != "no phone numbers":
        print(result)
```
